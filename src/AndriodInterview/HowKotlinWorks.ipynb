What is the difference between safe calls(?.) and null check(!!)?
Safe call operator i.e. ?. is used to check if the value of the variable is null or not. If it is null then null will be returned otherwise it will return the desired value.

var name: String? = "MindOrks"
println(name?.length) // 8
name = null
println(name?.length) // null

If you want to throw NullPointerException when the value of the variable is null, then you can use the null check or !! operator.

var name: String? = "MindOrks"
println(name?.length) // 8
name = null
println(name!!.length) // KotlinNullPointerException

1 What are the four main components of Android (Activity, Services, Broadcast Receiver, Content Provider)

An Activity represents a single screen with a user interface (like a login screen or a list screen). It handles user interactions and is the main entry point for users into the app.

A Service runs in the background without a UI to perform long‑running tasks such as playing music, downloading files, or syncing data. It keeps working even if the user leaves the screen.

A Broadcast Receiver listens for system‑wide or app‑specific broadcast messages (intents), such as battery‑low alerts, network‑change notifications, or custom events, and responds accordingly.

A Content Provider manages shared structured data (such as contacts, media, or app data) and allows other apps to securely query or modify that data using standard operations like insert, query, update, and delete.

Explain Activity Lifecycle with an example

The Activity lifecycle in Android defines how an activity moves through different states (created, running, paused, stopped, destroyed) and when the system calls specific callback methods like onCreate(), onStart(), onResume(), onPause(), onStop(), and onDestroy().

Main lifecycle callbacks
Here are the core methods you override in an Activity subclass:

onCreate() – Called once when the activity is first created; you set up the layout (setContentView) and initialize data.

onStart() – Called when the activity becomes visible to the user but not yet interactive.

onResume() – Called when the activity is about to start interacting with the user (foreground, fully active).

onPause() – Called when the activity loses focus (e.g., a dialog or new activity appears); you should pause animations, save UI‑state, or release some resources.

onStop() – Called when the activity is no longer visible; you release heavier resources (camera, sensor listeners, etc.).

onDestroy() – Called before the activity is destroyed (by user closing app or system killing it); you perform final cleanup.

States in order:
Created → Started → Resumed → Paused → Stopped → Destroyed.

Example with a simple login screen

Imagine a LoginActivity in your app:

When the app starts:

onCreate() → inflate login layout, bind EditText and Button.

onStart() → activity becomes visible (but not necessarily user‑focused yet).

onResume() → activity is fully active and ready for login‑button clicks.

When the user taps “Forgot Password?” and ForgotPasswordActivity opens:

LoginActivity.onPause() → you pause any background animations or save unsaved text.

ForgotPasswordActivity.onCreate() → onStart() → onResume() are called in sequence.

When the user presses Back from ForgotPasswordActivity:

ForgotPasswordActivity.onPause() → onStop() → onDestroy() (if finished).

LoginActivity.onRestart() → onStart() → onResume() → user returns to the login screen.

If the user hits the home button or the app is killed by the system:

LoginActivity.onPause() → onStop() → onStop marks that the activity is no longer visible.

Later, if the system restores it, it will restart from onCreate() (or onRestart() if it was just stopped).

In short, the Activity lifecycle ensures your UI and resources are set up when the screen appears and cleaned up when it is hidden or destroyed, while an example login flow shows how methods like onCreate(), onResume(), onPause(), and onStop() are fired as the user navigates between screens
